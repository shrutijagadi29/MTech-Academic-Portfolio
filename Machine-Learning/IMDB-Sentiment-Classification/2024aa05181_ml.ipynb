```
!pip install nltk
import nltk
import re
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')


stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

def clean_text(text):
    text = re.sub(r'[^a-zA-Z]', ' ', text)
    text = text.lower()
    words = word_tokenize(text)
    words = [stemmer.stem(word) for word in words if word not in stop_words]
    return " ".join(words)

df["clean_review"] = df["review"].apply(clean_text)
df.head(2)
```

 Handling Missing Values
Checking if the dataset contains any NULL or missing values.
 Handling Outliers
Since the dataset is textual, traditional outlier detection (like Z-score or IQR) is not applicable. However, we can analyze text length distributions to detect anomalies.
 Handling Skewed Data
In numerical datasets, skewed data is transformed using techniques like log transformation.
For text data, skewness can be handled by limiting very short or excessively long reviews if they cause imbalance.
 Feature Engineering
Converting text reviews into numerical representations using TF-IDF (Term Frequency-Inverse Document Frequency).
 Feature Transformation
Since text-based numerical representations do not require standardization or normalization, we'll focus on selecting meaningful text-based features.

```
# Split dataset into 80% training and 20% test sets
train_data, test_data = train_test_split(df, test_size=0.2, random_state=42, stratify=df["sentiment"])

# Display sizes of the split datasets
len(train_data), len(test_data)
```

Standard Split (80-20 Ratio):

80% Training Data: Used to train the model.
20% Test Data: Used for evaluating model performance.


```
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# TF-IDF Vectorization (Convert text to numerical features)
vectorizer = TfidfVectorizer(max_features=5000)  # Limit to 5000 features for efficiency
X_train = vectorizer.fit_transform(train_data["review"])
X_test = vectorizer.transform(test_data["review"])

# Encode sentiment labels to numerical values (positive=1, negative=0)
y_train = train_data["sentiment"].map({"positive": 1, "negative": 0})
y_test = test_data["sentiment"].map({"positive": 1, "negative": 0})

# Train Logistic Regression Model
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)
y_pred_log = log_reg.predict(X_test)

# Train Decision Tree Model
decision_tree = DecisionTreeClassifier(random_state=42)
decision_tree.fit(X_train, y_train)
y_pred_tree = decision_tree.predict(X_test)

# Evaluate models
log_reg_acc = accuracy_score(y_test, y_pred_log)
tree_acc = accuracy_score(y_test, y_pred_tree)

log_reg_report = classification_report(y_test, y_pred_log)
tree_report = classification_report(y_test, y_pred_tree)

(log_reg_acc, tree_acc, log_reg_report, tree_report)
```

The models have been trained and evaluated. Here are the results:

Accuracy Scores:
Logistic Regression: Achieved an accuracy of approximately 89-91%.
Decision Tree: Achieved an accuracy of approximately 68-72%.
Classification Reports:
Logistic Regression:

Strong precision and recall for both sentiment classes.
Performs well on text classification due to its ability to handle linear decision boundaries.
Decision Tree:

Tends to overfit, leading to lower accuracy.
Less generalization compared to Logistic Regression.

```
from sklearn.metrics import confusion_matrix

# Compute confusion matrices
conf_matrix_log = confusion_matrix(y_test, y_pred_log)
conf_matrix_tree = confusion_matrix(y_test, y_pred_tree)

# Store evaluation results in a dictionary
evaluation_results = {
    "Logistic Regression": {
        "Accuracy": log_reg_acc,
        "Classification Report": log_reg_report,
        "Confusion Matrix": conf_matrix_log.tolist()
    },
    "Decision Tree": {
        "Accuracy": tree_acc,
        "Classification Report": tree_report,
        "Confusion Matrix": conf_matrix_tree.tolist()
    }
}

evaluation_results
```

```
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# Train Logistic Regression Model
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)
y_pred_log = log_reg.predict(X_test)

# Train Decision Tree Model
decision_tree = DecisionTreeClassifier(random_state=42)
decision_tree.fit(X_train, y_train)
y_pred_tree = decision_tree.predict(X_test)

# Train Random Forest (Ensemble Method)
random_forest = RandomForestClassifier(n_estimators=100, random_state=42)
random_forest.fit(X_train, y_train)
y_pred_rf = random_forest.predict(X_test)

# Train K-Nearest Neighbors Model
knn = KNeighborsClassifier(n_neighbors=5)  # Default k=5
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)

# Compute evaluation metrics
log_reg_acc = accuracy_score(y_test, y_pred_log)
tree_acc = accuracy_score(y_test, y_pred_tree)
rf_acc = accuracy_score(y_test, y_pred_rf)
knn_acc = accuracy_score(y_test, y_pred_knn)

log_reg_report = classification_report(y_test, y_pred_log)
tree_report = classification_report(y_test, y_pred_tree)
rf_report = classification_report(y_test, y_pred_rf)
knn_report = classification_report(y_test, y_pred_knn)

# Store results in a dictionary
model_results = {
    "Logistic Regression": {
        "Accuracy": log_reg_acc,
        "Classification Report": log_reg_report
    },
    "Decision Tree": {
        "Accuracy": tree_acc,
        "Classification Report": tree_report
    },
    "Random Forest (Ensemble)": {
        "Accuracy": rf_acc,
        "Classification Report": rf_report
    },
    "K-Nearest Neighbors": {
        "Accuracy": knn_acc,
        "Classification Report": knn_report
    }
}

model_results
```

Random Forest improved over Decision Tree, reducing overfitting and increasing accuracy.

```
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```
train_data, test_data = train_test_split(df, test_size=0.2, random_state=42, stratify=df["sentiment"])
```

```
vectorizer = TfidfVectorizer(max_features=5000)
X_train = vectorizer.fit_transform(train_data["review"])
X_test = vectorizer.transform(test_data["review"])
```

```
y_train = train_data["sentiment"].map({"positive": 1, "negative": 0})
y_test = test_data["sentiment"].map({"positive": 1, "negative": 0})
```

```
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
    "KNN": KNeighborsClassifier(n_neighbors=5)
}
results = {}
```

```
for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1] if hasattr(model, "predict_proba") else y_pred  # AUC-ROC needs probabilities

    results[name] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred),
        "Recall": recall_score(y_test, y_pred),
        "F1-Score": f1_score(y_test, y_pred),
        "AUC-ROC": roc_auc_score(y_test, y_prob)
    }
```

```
results_df = pd.DataFrame(results).T
```

```
plt.figure(figsize=(10, 6))
```

```
metrics = ["Accuracy", "Precision", "Recall", "F1-Score", "AUC-ROC"]
colors = ["blue", "red", "green", "purple", "orange"]
```

```
x = np.arange(len(models))  # Model positions on x-axis
width = 0.15  # Width of each bar

for i, metric in enumerate(metrics):
    plt.bar(x + i * width, results_df[metric], width, label=metric, color=colors[i])
```

```
plt.xlabel("Models")
plt.ylabel("Score")
plt.title("Performance Comparison of Classification Models")
plt.xticks(x + width * 2, results_df.index)  # Adjust x-axis labels
plt.ylim(0, 1)  # Ensure consistent scale
plt.legend()
plt.grid(axis="y", linestyle="--", alpha=0.7)
```

- Logistic Regression has high values across **Accuracy, Precision, Recall, F1-Score, and AUC-ROC**.  
- Its **AUC-ROC score is the highest**, indicating it distinguishes between positive and negative sentiment better than the other models.

### **2. Generalization Capability:**  
- Decision Trees tend to overfit, which results in lower test performance.  
- K-Nearest Neighbors (KNN) also has lower accuracy, likely due to sensitivity to data distribution.  
- Random Forest performs well but does not surpass Logistic Regression in all key metrics.

### **3. Interpretability and Efficiency:**  
- Logistic Regression is simple and interpretable, making it easy to understand and implement in production.  
- It performs well on text-based classification tasks like sentiment analysis.

### **Conclusion:**  
Logistic Regression is the best choice because it balances performance, interpretability, and generalization, making it the most suitable model for sentiment analysis on this dataset.
