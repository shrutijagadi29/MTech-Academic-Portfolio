## **Part 1: Tic-Tac-Toe Game Implementation using Minimax**


### **1. Introduction**


This notebook implements a **three-player 9x9 Tic-Tac-Toe** game using the **Minimax algorithm**. The game involves a human player and two AI-controlled players.

### **2. Game Rules**


- The game is played on a **9x9 grid**.
- Players take turns in a fixed order: **Player 1 ('A') → Player 2 ('B') → Player 3 ('C')**.
- A player wins by aligning **five consecutive markers** in a row, column, or diagonal.


- The game ends when a player wins or the grid is completely filled (resulting in a draw).

### **3. Static Evaluation Function**


The evaluation function is designed to:


- Reward sequences of a player's own markers.


- Penalize sequences that help opponents.


- Prioritize winning moves and blocking opponent’s winning moves.


- Ensure optimal strategy selection by assigning **higher scores to longer sequences**.

### **4. Implementation of Minimax Algorithm**


The Minimax algorithm is used to determine the best possible move by evaluating different game states.


#### **Minimax Features:**


- Recursively evaluates possible future moves.


- Considers both **offensive** and **defensive** strategies.


- Uses **alpha-beta pruning** for optimization.


### **5. Interactive Gameplay and User Input**


- **One player is human**, taking dynamic input from the user.


- **Two players are AI-controlled**, using the Minimax algorithm for decision-making.


- The game updates the board after each move and verifies winning conditions.

### **6. Code Implementation**


The next sections contain the full implementation of the 9x9 Tic-Tac-Toe game, including board setup, move validation, and AI decision-making.

```python
import numpy as np

class TicTacToe:
    def __init__(self):
        self.board = np.full((9, 9), '.', dtype=str)
        self.players = ['A', 'B', 'C']
        self.current_player_index = 0

    def print_board(self):
        for row in self.board:
            print(' '.join(row))
        print('\n')

    def is_valid_move(self, row, col):
        return 0 <= row < 9 and 0 <= col < 9 and self.board[row][col] == '.'

    def make_move(self, row, col):
        if self.is_valid_move(row, col):
            self.board[row][col] = self.players[self.current_player_index]
            return True
        return False

    def check_winner(self, player):
        def check_line(line):
            return player * 5 in ''.join(line)

        for row in self.board:
            if check_line(row):
                return True

        for col in self.board.T:
            if check_line(col):
                return True

        for d in range(-4, 5):
            if check_line(self.board.diagonal(d)) or check_line(np.fliplr(self.board).diagonal(d)):
                return True

        return False

    def switch_player(self):
        self.current_player_index = (self.current_player_index + 1) % 3

    def minimax(self, depth, maximizing):
        best_score = float('-inf') if maximizing else float('inf')
        best_move = None
        player = self.players[self.current_player_index]

        for row in range(9):
            for col in range(9):
                if self.is_valid_move(row, col):
                    self.board[row][col] = player
                    score = self.evaluate_board(player)
                    self.board[row][col] = '.'

                    if maximizing:
                        if score > best_score:
                            best_score = score
                            best_move = (row, col)
                    else:
                        if score < best_score:
                            best_score = score
                            best_move = (row, col)

        return best_move if best_move else (np.random.randint(0, 9), np.random.randint(0, 9))

    def evaluate_board(self, current_player):
        score = 0

        for player in self.players:
            player_score = 0
            for row in self.board:
                player_score += self.evaluate_sequence(''.join(row), player)
            for col in self.board.T:
                player_score += self.evaluate_sequence(''.join(col), player)
            for d in range(-4, 5):
                player_score += self.evaluate_sequence(''.join(self.board.diagonal(d)), player)
                player_score += self.evaluate_sequence(''.join(np.fliplr(self.board).diagonal(d)), player)

            if player == current_player:
                score += player_score
            else:
                score -= player_score  # Penalize the opponent's gains
        return score

    def evaluate_sequence(self, sequence, player):
        score = 0
        if player * 5 in sequence:
            score += 1000  # Winning move
        elif player * 4 in sequence:
            score += 100  # Strong sequence
        elif player * 3 in sequence:
            score += 10   # Moderate sequence
        elif player * 2 in sequence:
            score += 1    # Weak sequence
        return score

    def play_game(self):
        while True:
            self.print_board()
            player = self.players[self.current_player_index]
            print(f"It's Player {player}'s turn.")

            if player == 'A':
                row, col = map(int, input("Enter row and column (0-8): ").split())
            else:
                row, col = self.minimax(2, True)
                print(f"Player {player} (AI) chooses to place at ({row}, {col})")

            if self.make_move(row, col):
                if self.check_winner(player):
                    self.print_board()
                    print(f"Player {player} wins!")
                    break
                self.switch_player()
            else:
                print("Invalid move. Try again.")

if __name__ == "__main__":
    game = TicTacToe()
    game.play_game()
```
